### Measure of Dispersion

1. Variance
2. Standard Deviation

> Why we need Variance and SD instead of mean/median: mean and median describe the center of a distribution but do not show how data are spread around that center. Two datasets can have the same mean yet very different variability:
- Left (low spread): {20, 21, 22, 23, 24} — mean = 22, SD ≈ 1.58  
- Right (high spread): {2, 12, 22, 32, 42} — mean = 22, SD ≈ 14.14

- Variance (σ²) = average of squared deviations from the mean: σ² = Σ(xᵢ − μ)² / N  , i=1 to N (Population size)
    - For sample data, use sample variance: S² = Σ(xᵢ − x̄)² / (n−1), i=1 to n. The denominator is n−1 because the sample mean x̄ is estimated from the same data, so one degree of freedom is lost(or it is n-1 coz, there is one paramaeter i.e, mean, that is overestimated ). This is Bessel’s correction, and it corrects the downward bias that occurs if you divide by n.

    - Example: sample {10, 12, 14}. Mean x̄ = 12. Squared deviations: (10−12)²=4, (12−12)²=0, (14−12)²=4, total=8.
        - Population variance would be 8/3 ≈ 2.67,
        - Sample variance is 8/(3−1) = 4.
      The sample formula gives an unbiased estimate of the true population variance, which is what you need in statistics and modeling.
- Standard deviation (σ) = √variance — returns dispersion in the same units as the data  
- Interpretation: small SD → values cluster tightly around the mean; large SD → values are widely dispersed  
- Practical notes: variance/SD are sensitive to outliers and are essential for measuring consistency, comparing datasets, and building statistical models (confidence intervals, hypothesis tests).

Histogram

## Key Concepts
- **Purpose**: Visualizes frequency distribution of continuous data
- **Bins**: Divide data into ranges; more bins provide greater detail
- **Use Cases**: Analyze distributions, identify patterns, detect outliers

## Interview-Ready Insights
- **Skewness Detection**: Reveals if data is left-skewed, right-skewed, or symmetric
- **Normality Assessment**: Identifies if data follows a normal (bell curve) distribution
- **Data Concentration**: Shows where most values cluster and identifies multimodal distributions
- **Outlier Identification**: Sparse bars at extremes indicate potential outliers
- **Business Application**: Common in quality control, performance analysis, and risk assessment

## Best Practices
- Choose bin size carefully—too few bins hide patterns, too many create noise
- Label axes clearly with units
- Include frequency counts or percentages for context

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data: ages of 1000 people
ages = np.random.normal(loc=35, scale=15, size=1000)
ages = ages[(ages > 0) & (ages < 100)]  # Filter realistic ages

# Histogram Genmration
plt.figure(figsize=(10, 6))
plt.hist(ages, bins=30, color='skyblue', edgecolor='black', alpha=0.7)
plt.xlabel('Age', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.title('Distribution of Ages - Histogram', fontsize=14, fontweight='bold')
plt.grid(axis='y', alpha=0.3)
plt.show()
```
