# Types of Variables

## 1. Quantitative Variables
Numerical data that can be measured or counted.

### a. Discrete
- Takes countable, distinct values (whole numbers)
- Cannot take fractional values
- Examples: Age, number of students, count of defects
- Used with: Probability Mass Function (PMF)

### b. Continuous
- Takes any value within a range (can be fractional)
- Measured on a scale
- Examples: Height, weight, temperature, time
- Used with: Probability Density Function (PDF)

## 2. Qualitative (Categorical) Variables
Non-numerical data that describes qualities or characteristics.
- Examples: Gender (M/F), Color (red, green, blue), Product category
- Cannot be used directly in mathematical operations
- Requires encoding for statistical analysis

**Interview Tip:** Understanding the distinction between discrete, continuous, and categorical variables is essential for:
- Selecting appropriate statistical methods
- Choosing correct visualization techniques
- Designing data collection strategies

## Random Variables

A random variable is a function that assigns numerical values to the outcomes of a random experiment.

**Types of Random Variables:**

1. **Discrete Random Variable**
    - Takes countable, distinct values
    - Example: Number of coin flips until heads appears, number of students in a class
    - Probability described by Probability Mass Function (PMF)

    ex : Tossing of coin 
    x = {
        H , 0
        T, 1
    }

2. **Continuous Random Variable**
    - Takes any value within a range
    - Example: Height, temperature, time
    - Probability described by Probability Density Function (PDF)

    ex : Tommorow how many inches it is goiing to rain
    0, 1.1, 5.4, 6.4

**Key Concepts:**
- **Probability Distribution**: Shows all possible values and their probabilities
- **Expected Value (Mean)**: Long-run average of the random variable
- **Variance**: Measure of spread around the mean
- **Standard Deviation**: Square root of variance

**Interview Tip:** Random variables form the foundation of probability theory and statistics. Understanding the distinction between discrete and continuous types is crucial for selecting appropriate statistical methods.
