# Percentile: A Value-Based Measure

**Definition:** A percentile is a value below which a certain percentage of observations lie.

**Example Dataset:** {2, 2, 3, 4, 5, 5, 6, 7, 8, 8, 8, 9, 9, 10}

## Calculating Percentile Rank

To find the percentile rank of a specific value:

$$\text{Percentile Rank} = \frac{\text{Number of values below target}}{n} \times 100$$

**Example:** Percentile rank of 9 = (11 ÷ 14) × 100 = **78.57th percentile**

## Finding the Value at a Given Percentile

To find the value at a specific percentile (e.g., 25th percentile):

$$\text{Position} = \frac{P}{100} \times (n + 1)$$

Where P is the percentile and n is the number of observations.

**Example:** 25th percentile of the dataset:
- Position = (25 ÷ 100) × 15 = **3.75**
- Since 3.75 falls between indices 3 and 4, interpolate: (3 + 4) ÷ 2 = **3.5**

## Key Takeaway

The 25th percentile value is 3.5, meaning **25% of the distribution lies below this value**.

---

**Interview Tip:** Percentiles divide data into 100 equal parts; quartiles divide into 4 parts (25th, 50th, 75th percentiles).


## Quartiles: The 4-Way Data Split

**Definition:** Quartiles divide a dataset into 4 equal parts, each containing 25% of the data.

| Quartile | Percentile | Meaning |
|----------|-----------|---------|
| Q1 | 25th | 25% of data falls below this value |
| Q2 | 50th | Median - 50% of data falls below this value |
| Q3 | 75th | 75% of data falls below this value |
| Q4 | 100th | Maximum value |

**Interquartile Range (IQR):** Q3 - Q1 (contains middle 50% of data, used for outlier detection)

**Interview Tip:** Quartiles are specific percentiles that partition data into 4 equal sections. If a value is at Q1, it means 75% of the data is above it and 25% is below it.


# Five-Number Summary

A five-number summary provides a quick statistical overview of a dataset:

| Component | Description |
|-----------|-------------|
| **Minimum** | Smallest value (Q0) |
| **Q1 (25th Percentile)** | 25% of data falls below this value |
| **Median (Q2, 50th Percentile)** | Middle value; divides dataset in half |
| **Q3 (75th Percentile)** | 75% of data falls below this value |
| **Maximum** | Largest value (Q4) |

## Example: {1, 2, 2, 2, 3, 3, 4, 5, 5, 5, 6, 6, 6, 6, 7, 8, 8, 9, 27}

lower fence (anything lower than this will remove as outlier)<---------> Higher Fence (anything higher than this value is removed as outlier)

lower fence = Q1 - 1.5*IQR
higher fence = Q3 + 1.5 * IQR

## Key Insights

- **Interquartile Range (IQR):** Q3 - Q1 = 4 (middle 50% of data)
- **Outlier Detection:** Values > Q3 + 1.5×IQR or < Q1 - 1.5×IQR are outliers (e.g., 27 is an outlier)
- **Skewness:** Large gap between median and max suggests right-skewed distribution

**Interview Tip:** Five-number summary is the foundation for box plots and outlier detection in data analysis.
