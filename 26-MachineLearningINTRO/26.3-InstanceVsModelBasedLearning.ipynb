## Instance Based Learning vs Model Based Learning

## Instance Based Learning
- Learns by storing training examples and comparing new inputs to them
- No explicit parametric model; prediction uses similarity or distance to stored cases
- Training is cheap, prediction is expensive; best for problems with local decision structure
- Common approaches: k-NN, case-based reasoning, memory-based collaborative filtering
- Strengths: adapts to new data without retraining; works well when patterns vary locally
- Weaknesses: high memory use, slower inference, sensitive to feature scaling

Examples:
- k-NN classification/regression
- Case-based reasoning in diagnostics or help desks
- Memory-based collaborative filtering in recommender systems

## Model Based Learning
- Learns a parametric or nonparametric approximation of the underlying data distribution
- Builds a model from training data, then uses the model to make predictions
- Training can be expensive; inference is typically fast
- Best when the problem has a stable global pattern and large datasets

Interview note:
- Highlight that model-based methods are “eager learners”: they generalize during training.
- Common tradeoff is bias versus variance depending on model complexity.

Examples:
- Linear regression, logistic regression
- Decision trees, SVM, neural networks
- Probabilistic models like Naive Bayes






- Instance-based learning
    - The system keeps the training examples.
    - For a new input, it compares that input to stored examples and uses similarity or distance to decide.
    - It does not build a compact global model in training.
    - Training is simple, but prediction can be slower.
    - Good when the decision depends on local patterns around each query.
    - Memorizing the Data

- Model-based learning
    - The system builds a model from the training data.
    - That model generalizes and is used for future predictions.
    - Training can be more expensive, but inference is usually faster.
    - It works well when there is a stable global relationship in the data.
    - Here model learn the pattern of the data.
