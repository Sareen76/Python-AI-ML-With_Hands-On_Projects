## Geometric Interpretation of Weight Vector

> **Key Insight**: The weight vector **W** is always perpendicular to the decision boundary (hyperplane).

### Why is W perpendicular to the hyperplane?

For any point on the hyperplane, the equation W^T · x + b = 0 holds. If we take two different points x₁ and x₂ on this hyperplane:
- W^T · x₁ + b = 0
- W^T · x₂ + b = 0

Subtracting these equations: W^T · (x₁ - x₂) = 0

Since (x₁ - x₂) is a vector lying on the hyperplane, and W^T · (x₁ - x₂) = 0, this means **W is orthogonal (perpendicular) to every vector on the hyperplane**.

### The Dot Product and Angle θ

In the formula |W| · |x| · cos θ:
- When θ = 90° (perpendicular): cos(90°) = 0, so W^T · x = 0
- When θ = 0° (parallel): cos(0°) = 1, so W^T · x = |W| · |x| (maximum)

This is why **W must be perpendicular to the decision boundary** — because on the boundary itself, W^T · x + b = 0, meaning W and x are orthogonal.

### 3D to 2D Projection

In 3D space, a hyperplane (2D surface) is defined by the equation W^T · x + b = 0. The weight vector W points in the direction **normal** (perpendicular) to this plane. When viewing from above, you see the 2D line where the plane intersects your view — W still points perpendicular to this line.
