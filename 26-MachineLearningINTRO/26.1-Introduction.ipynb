## Difference between AI, ML, Deep Learning, Data Science

### Short definitions
- **Artificial Intelligence (AI)**: Broad field focused on creating systems that perform tasks that normally require human intelligence (planning, reasoning, perception, language).
- **Machine Learning (ML)**: Subset of AI that enables systems to learn patterns from data and make predictions without explicit programming.
- **Deep Learning (DL)**: Subset of ML using multi-layer (deep) neural networks to learn hierarchical feature representations from raw data.
- **Data Science**: Interdisciplinary practice of extracting insights from data using statistics, ML, visualization, and domain knowledge; includes data engineering and storytelling.

---

### Key distinctions (concise)
- Scope: AI > ML > DL (DL is a technique inside ML).
- Data needs: DL >> ML > classical AI rules.
- Feature engineering: ML often needs manual features; DL learns features automatically.
- Compute: DL typically requires GPUs and more compute.
- Interpretability: Classical ML models (linear, tree) are more interpretable than most DL models.
- Use cases: Rule-based systems (AI), regression/classification (ML), image/audio/NLP at scale (DL), product insights/experiments (Data Science).

---

### Short examples
- AI: A rules-based chat bot that follows scripted decision trees.
- ML: Random Forest to predict churn from customer features.
- DL: Convolutional Neural Network for image classification.
- Data Science: A/B test analysis, feature engineering, dashboard + ML model for conversion optimization.

---

### Interview-level notes & succinct Q&A
- Q: How do ML and DL differ in practice?  
    A: DL uses neural nets to learn features end-to-end and needs more data/compute; ML uses explicit features and simpler models (SVM, tree, LR).

- Q: When choose DL vs classical ML?  
    A: Choose DL for unstructured data (images, audio, text) and large datasets; choose classical ML for tabular data, interpretability, or small datasets.

- Q: What is overfitting and how to prevent it?  
    A: Model fits noise → use more data, regularization (L1/L2, dropout), simpler model, cross-validation, early stopping.

- Q: Explain bias-variance tradeoff briefly.  
    A: Bias = error from wrong assumptions (underfit); Variance = error from sensitivity to training data (overfit). Balance via model complexity and regularization.

- Q: Important evaluation metrics?  
    A: Classification: Accuracy, Precision, Recall, F1, ROC-AUC. Regression: MSE, RMSE, MAE, R². Use business-aware metrics and confusion matrix for imbalance.

- Q: Role of feature engineering?  
    A: Critical for classical ML—creates signal for models. In DL, less manual engineering but preprocessing still matters.

- Q: How to handle imbalanced classes?  
    A: Resampling (oversample/undersample), synthetic samples (SMOTE), class-weighted loss, proper metrics (precision/recall).

- Q: Deployment considerations?  
    A: Model size/latency, monitoring (drift, data quality), CI/CD for models, reproducibility, inference infrastructure (CPU/GPU), privacy/compliance.

- Q: Fast way to validate a model idea?  
    A: Start with simple baseline (logistic regression/decision tree), cross-validate, measure lift vs baseline, then iterate.

- Q: Common pitfalls in production ML/DL?  
    A: Data leakage, training/serving skew, insufficient monitoring, ignoring concept drift, overfitting to validation, poor feature versioning.

### Quick code concept (conceptual)
- Classical ML (scikit-learn): featurize → fit simple model → evaluate.  
- DL (PyTorch/TensorFlow): raw input (images/text) → neural net → train on GPUs → evaluate.

## 3 Types of Machine Learning Techniques

### 1. Supervised Learning
- Definition: Learn a mapping from inputs to outputs using labeled data.  
- Common tasks & algorithms: Classification (Logistic Regression, Random Forest, SVM), Regression (Linear Regression, XGBoost).  
- Examples: Email spam detection, house price prediction, image classification.  
- When to choose: You have reliable labels and want direct predictions or probabilities.

### 2. Unsupervised Learning
- Definition: Discover patterns or structure from unlabeled data.  
- Common tasks & algorithms: Clustering (K-Means, DBSCAN), Dimensionality Reduction (PCA, t-SNE), Density Estimation.  
- Examples: Customer segmentation, anomaly detection, feature extraction for downstream models.  
- When to choose: No labels available or you want exploratory insights / representations.

### 3. Reinforcement Learning (RL)
- Definition: Learn a policy that maximizes cumulative reward through interaction with an environment.  
- Common algorithms: Q-Learning, DQN, Policy Gradients, PPO.  
- Examples: Game playing (AlphaGo), robotics control, recommendation with sequential interaction.  
- When to choose: Sequential decision problems with delayed feedback and clear reward signal.

## Interview-level Notes (concise)
- Supervised vs Unsupervised: labeled vs unlabeled data; supervised predicts, unsupervised discovers structure.  
- Metrics to cite: Classification → Accuracy, Precision/Recall, F1, ROC-AUC. Regression → MSE, MAE, R². Clustering → Silhouette, Davies–Bouldin.  
- Handling imbalanced data: resample (oversample/undersample), SMOTE, class weights, appropriate metrics (precision/recall, PR-AUC).  
- Overfitting signs & fixes: large train/validation gap → regularize (L1/L2, dropout), simplify model, more data, cross-validation, early stopping.  
- Feature engineering: critical for tabular supervised ML; transformations, interaction features, encoding categorical variables.  
- When to use DL vs classical ML: DL for large unstructured data (images, text, audio); classical ML for smaller/tabular data and interpretability.  
- RL vs supervised: RL needs environment & reward; not appropriate if only static labeled dataset exists.  
- Quick practical tip: start with a simple baseline (logistic regression / decision tree), validate via cross-validation, then iterate complexity only if it improves business-relevant metrics.


### Supervised Learning

- **Regression**
    - Predicts continuous values
    - Examples: house price prediction, sales forecasting, demand estimation
    - Typical algorithms: linear regression, ridge/lasso, random forest regression, gradient boosting

- **Classification**
    - Predicts discrete labels
    - Types:
        - **Binary**: two classes
            - Examples: spam vs not spam, fraud vs legit, churn vs retain
        - **Multiclass**: more than two classes
            - Examples: digit recognition, topic labeling, product category prediction
    - Typical algorithms: logistic regression, decision trees, random forest, SVM, XGBoost

### Interview-level notes

- Use regression when the target is numeric and ordered; use classification when the target is categorical.
- Use binary classification for yes/no decisions; use multiclass when multiple categories matter.
- Common regression algorithms:
    - linear regression for simple, interpretable relationships
    - ridge/lasso for regularization and feature selection
    - tree-based regressors for nonlinear patterns
- Common classification algorithms:
    - logistic regression for baseline and interpretable scores
    - random forest for robust performance on mixed data
    - SVM/XGBoost for higher accuracy on structured data

- Example question: “How would you handle imbalanced classes?”
    - Answer: resampling (oversample minority / undersample majority), use class weights, choose metrics like precision/recall, F1, or PR-AUC instead of accuracy.

- Example question: “When does regression fail?”
    - Answer: when the target is categorical, when the relationship is not well represented by numeric distance, or when class boundaries matter more than error magnitude.

- Practical note:
    - Start with a simple baseline model and clear evaluation metric.
    - If the problem is tabular and labels exist, supervised learning is usually the right choice.
    - For imbalanced or rare-event problems, focus on business-relevant metrics and data sampling strategies.

### Unsupervised ML

- Definition: Learn patterns or structure from unlabeled data (without any target feature like no result ) without explicit target values.
- Common tasks:
    - Clustering: group similar items (K-Means, DBSCAN, hierarchical clustering)
    - Dimensionality reduction: compress features while preserving structure (PCA, t-SNE, UMAP)
    - Anomaly detection: find rare or unusual observations (isolation forest, one-class SVM)
- Use cases:
    - Customer segmentation from purchase behavior
    - Topic discovery from text corpora
    - Feature extraction for downstream supervised models
    - Detecting fraud or equipment failure when labels are scarce

### Interview-level notes

- When to choose unsupervised learning:
    - No labels available
    - You need exploratory insights, data structure, or representation learning
    - You want to find groups, reduce dimensionality, or detect outliers

- Practical examples:
    - Clustering: segment customers into groups for targeted marketing
    - PCA: reduce thousands of features to a few components for visualization or noise reduction
    - Anomaly detection: identify unusual transactions in financial data

- Common pitfalls:
    - Interpretation is harder because there is no ground truth
    - Evaluation often relies on domain knowledge or proxy metrics
    - Clusters may not correspond to meaningful business groups

- Interview talking points:
    - “Unsupervised learning discovers structure in data, while supervised learning maps inputs to known outputs.”
    - “Use clustering for segmentation and dimensionality reduction for visualization or preprocessing.”
    - “Evaluate clustering with silhouette score, Davies-Bouldin index, or by validating clusters against known business outcomes.”
    - “Anomaly detection is valuable when labeled anomalies are rare or unavailable, but you must verify false positives carefully.”

- Quick tip:
    - Start with simple techniques like K-Means or PCA, inspect results visually, then refine with more advanced methods if needed.

## Reinforcement Learning

- Definition: Learn a policy to make decisions by interacting with an environment and maximizing cumulative reward.
- Core concept: agent, environment, state, action, reward.
- Typical setting: sequential decision-making where current action affects future states and rewards.

### Common algorithms
- Value-based: Q-Learning, Deep Q-Networks (DQN)
- Policy-based: REINFORCE, PPO, A2C
- Actor-Critic: combines value and policy learning

### Practical examples
- Game playing: board games, Atari, chess, Go
- Robotics: motion control, grasping, navigation
- Recommendation: sequential recommendations with feedback over time
- Resource allocation: scheduling, inventory management

### Interview-level notes
- When to choose RL:
    - problem involves sequential decisions
    - reward is delayed
    - environment dynamics matter
- When not to choose RL:
    - only static labeled data exists
    - immediate prediction is enough
    - exploration is unsafe or too expensive
- Key tradeoffs:
    - exploration vs exploitation
    - sample efficiency vs model complexity
    - stability of training and reward design
- Evaluation:
    - cumulative reward over episodes
    - task success rate
    - sample efficiency and robustness
- Quick answer:
    - “Use RL when the task is about learning a strategy through interaction and rewards, not when you just need a supervised mapping from inputs to labels.”
