# Nominal/OHE Encoding

- One-Hot Encoding (OHE) is used to convert nominal categorical variables into numeric form.
- It creates a separate binary feature for each category.
- Each row has a value of `1` in the column corresponding to its category and `0` elsewhere.

### Why it matters

- Machine learning models generally require numeric input.
- Nominal categories have no intrinsic order, so assigning arbitrary numbers can mislead models.
- OHE preserves category identity without introducing order.

### When to use

- Use for categorical features with no natural ranking, such as:
    - `Gender` (`Male`, `Female`, `Other`)
    - `Color` (`Red`, `Green`, `Blue`)
    - `Country` (`USA`, `India`, `Germany`)

### Interview-level notes

- OHE increases dimensionality; avoid it for high-cardinality features unless necessary.
- For tree-based models, label encoding may work, but OHE is safer for linear models.
- Watch out for multicollinearity: drop one dummy column (dummy variable trap) if using regression-based algorithms.
- Common implementation tools:
    - `pandas.get_dummies()`
    - `sklearn.preprocessing.OneHotEncoder()`

### Example

- Original feature: `Color = [Red, Blue, Red, Green]`
- One-hot encoded:
    - Red : [1, 0, 0]
    - Green: [0, 1, 0]
    - Blue :  [0, 0, 1]

This makes the categorical data usable for most ML pipelines while keeping the representation interpretable.

> The only Disadvantage is that for large categories or like 100 categories you will have to make 100 binary represnetation for each categories.

> Sparse Matrix will be formed ===> Which results in Overfitting 

```python
# Implementation of One hot Encoding using sklearn
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
# Create a sample DataFrame
data = {
    'Color': ['Red', 'Blue', 'Green', 'Red', 'Green']
    }
df = pd.DataFrame(data)
df.head()
```

```python
# Create an instance of OneHotEncoder
encoder = OneHotEncoder()

# Fit and transform the data
encoded_values = encoder.fit_transform(df[['Color']]).toarray()
encoded_values
```

```python
encoder_df = pd.DataFrame(encoded_values, columns=encoder.get_feature_names_out(['Color']))
encoder_df
```

```python
pd.concat([df, encoder_df], axis=1)
```

```python
import seaborn as sns
data = sns.load_dataset('tips')
data.head()


encoder = OneHotEncoder()
encode_data = encoder.fit_transform(data[['sex', 'smoker', 'day', 'time']]).toarray()
encoder_df = pd.DataFrame(encode_data, columns=encoder.get_feature_names_out(['sex', 'smoker', 'day', 'time']))
pd.concat([data, encoder_df], axis=1)
```

### Label Encoding

- Convert categories to integers (e.g., {Low:0, Medium:1, High:2}) — preserves order for ordinal features.
- Use when categories have a natural ranking (education level, rating bins, size).

When to use
- Ordinal features with meaningful order.
- Algorithms that can handle integer-encoded categories (tree-based models) or when you explicitly want to inject order.

Pitfalls & best practices
- Never arbitrarily label nominal categories — integers imply order and can mislead linear models.
- For nominal data prefer One-Hot or target/frequency/embedding encodings.
- Fit encoder on training data and reuse on validation/test to avoid leakage.
- Store/save mapping (dict or encoder object) for reproducibility and inverse_transform.
- Handle unseen categories explicitly (e.g., sklearn OrdinalEncoder handle_unknown or map to -1).
- Missing values: impute or reserve a sentinel value before encoding.
- High-cardinality: label encoding reduces dimensionality but may still harm linear models — consider target/frequency encoding or embeddings.

Interview tips (concise)
- Use sklearn.preprocessing.LabelEncoder for single-column labels; sklearn.preprocessing.OrdinalEncoder for feature columns.
- Explain why label encoding can inject ordinal bias and when that bias is acceptable (trees) vs harmful (linear/regression).
- Mention alternatives for high-cardinality features and how to handle unseen labels in production.

## Ordinal Encoding

- It is used to encode categorical features with a meaningful order.
- Each category is mapped to an integer based on its rank.

### When to use
- Use for ordinal features like:
    - Education: `High School < Bachelor < Master < PhD`
    - Size: `Small < Medium < Large`
    - Ratings: `Poor < Average < Good < Excellent`

### Why it matters
- Preserves the relative order of categories.
- Useful for models that can interpret numeric order, especially tree-based models.

### Example
- `Education = [High School, Bachelor, Master, PhD]`
- Ordinal encoded as:
    - High School: `0`
    - Bachelor: `1`
    - Master: `2`
    - PhD: `3`

### Interview-level notes
- Do not use ordinal encoding for nominal categories; it introduces false order.
- For linear models, ordinal encoding may still mislead if the distances between categories are not equal.
- Fit encoder only on training data and apply the same mapping to validation/test sets.
- Save the category-to-integer mapping for reproducibility and inference.
- Handle unseen categories explicitly (`unknown` category, `-1`, or custom mapping) to avoid runtime errors.
- Consider alternatives like target encoding or embeddings for high-cardinality ordered features.

```python
## Ordinal Encoding using sklearn
from sklearn.preprocessing import OrdinalEncoder

df = pd.DataFrame(
    {
        'size': ['S', 'M', 'L', 'M', 'S'],
    }
)
df.head()
```

```python
# Create an instance of OrdinalEncoder
ordinal_encoder = OrdinalEncoder(categories=[['S', 'M', 'L']]) # Keep the categiories in the order you want to encode them
ordinal_encoder.fit_transform(df[['size']])
ordinal_encoder.transform([['M']])



```

# Target Guided Ordinal Encoding

Target-guided ordinal encoding (a.k.a. target encoding) maps categorical levels to numeric values derived from the target variable (typically the mean target per category). It captures the relationship between a category and the target while reducing dimensionality compared to OHE.

Why it works
- Replaces categories with their target-statistics (mean, likelihood, log-odds), summarizing predictive signal.
- Useful for high-cardinality features where OHE is impractical.

How to compute (simple)
1. For each category, compute mean(target | category).
2. Optionally sort categories by mean and assign ordinal ranks.
3. Replace category with its statistic or rank.

Example
- Data: Color {Red,Blue,Green} with target mean {Red:0.2, Blue:0.7, Green:0.45}
- Encodings:
    - Mean encoding: Red→0.2, Blue→0.7, Green→0.45
    - Rank encoding: Red→0, Green→1, Blue→2

Best practices / interview-level notes
- Avoid target leakage:
    - Fit encodings only on training folds. Use K-fold (out-of-fold) or leave-one-out schemes to compute encodings for training.
    - For validation/test, map categories using training statistics; unseen categories -> global prior or sentinel value.
- Regularize/smooth estimates:
    - Blend category mean with global mean (weight by count) to stabilize rare categories.
    - Use additive smoothing or Bayesian priors.
- Use appropriate statistic:
    - For classification use mean target or log-odds; for regression use mean or median.
- Watch for overfitting:
    - High-cardinality categories with few samples can produce spurious high correlations; use smoothing and CV-based computation.
- Save mapping and pipeline:
    - Persist the category→value mapping and transform consistently in production.
- When to prefer alternatives:
    - Use OHE for low-cardinality nominal features where interpretability matters.
    - Consider frequency encoding, embedding, or hashing for very high-cardinality features.

Concise evaluation tips
- Compare model performance with and without target encoding via cross-validation.
- Monitor feature importance and stability across folds to detect overfitting introduced by encoding.

```python
import pandas as pd

df = pd.DataFrame(
    {
        'city' : ['Houston', 'Chicago', 'Chicago', 'Houston', 'Phoenix'],
        'price' : [1000, 800, 600, 700, 900]
    }
)

df.head()
```

```python
mean_price = df.groupby('city')['price'].mean().to_dict()
```

```python
df['city_encoded'] = df['city'].map(mean_price)
df
```

```python
# We will give only this two features to our Model and train it

df[['city', 'city_encoded']].head()
```
