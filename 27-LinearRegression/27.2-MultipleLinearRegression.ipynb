## Multiple Linear Regression Practical

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

```python
df_index = pd.read_csv("./economic_index.csv")
df_index.head()
```

```python
# Drop unnecssary columns
df_index.drop(columns=["Unnamed: 0", "year", "month"], inplace=True)
```

```python
df_index.isnull().sum()
```

```python
# Let's do Visualisation
import seaborn as sns
sns.pairplot(df_index)
```

```python
df_index.corr()
```

```python
## Visualize the dastapoints more closely

plt.scatter(df_index['interest_rate'], df_index['unemployment_rate'])
plt.xlabel("interest_rate")
plt.ylabel("unemployment_rate")
```

```python
## Get my independent and dependent Features

X = df_index[['interest_rate', 'unemployment_rate']]
# OR
X = df_index.iloc[:, :-1] # Take all the rows and leave the last column

X.head()
```

```python
Y = df_index.iloc[:,-1]
Y.head()
```

```python
# Train Test Split
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.25, random_state=42)
```

## sklearn standard scaling

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.fit_transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression
regression = LinearRegression()
regression.fit(X_train, Y_train)
```

```python
from sklearn.model_selection import cross_val_score
validation_score = cross_val_score(regression, X_train, Y_train, scoring="neg_mean_squared_error", cv=3)

# cv: number of folds used to split the training data for cross-validation.
# scoring: metric used to evaluate each fold; here we use negative mean squared error.
# cross_val_score trains on (cv-1)/cv of the data and validates on the remaining 1/cv portion,
# repeating this process cv times with different validation folds.
```

```python
validation_score
```

```python
np.mean(validation_score)
```

```python
# Prediction

Y_pred = regression.predict(X_test)
```

```python
Y_pred
```

```python
## Performance Metric

from sklearn.metrics import mean_absolute_error, mean_squared_error
```

```python
mse = mean_squared_error(Y_test, Y_pred)
mae = mean_absolute_error(Y_test, Y_pred)
rmse = np.sqrt(mse)
print ("MSE : ", mse)
print ("MAE : ", mae)
print ("RMSE : ", rmse)
```

```python
from sklearn.metrics import r2_score
score = r2_score(Y_test, Y_pred)
print(score)
```

```python
# Adjustted R2
1 - (1-score) * (len(Y_test)-1)/ (len(Y_test)-X_test.shape[1]-1)
```
