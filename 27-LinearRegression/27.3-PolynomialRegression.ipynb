# Polynomial Regression Practical

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
X = 6 * np.random.rand(100, 1) -3
Y = 0.5 * X**2 + 1.5*X + 2 + np.random.randn(100, 1)
# Quadratic equation used y = 0.5x^2 + 1.5x + 2 + outliers
plt.scatter(X, Y, color='g')
plt.xlabel('X Dataset')
plt.ylabel('Y Dataset')
```

```python
# Train Test Split

from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.24, random_state=42)
```

```python
# Train the Model
from sklearn.linear_model import LinearRegression
regression = LinearRegression()
```

```python
regression.fit(X_train, Y_train)
```

```python
# Performance Matrix

from sklearn.metrics import r2_score
score = r2_score(Y_test, regression.predict(X_test))
```

```python
score
```

# Let's Visualize the model

```python
plt.plot(X_train, regression.predict(X_train), color='r')
plt.scatter(X_train, Y_train)
plt.xlabel("X Dataset")
plt.ylabel("Y Dataset")
```

- The reason why our accuracy was too low , we need Polynomial Regression model for the prediction

```python
### POLYNOMIAL REGRESSION (htheta(x) = B0 + B1 x1)

# How to create the degree 1, 2 3 .... for the model

from sklearn.preprocessing import PolynomialFeatures

```

```python
poly = PolynomialFeatures(degree=2, include_bias=True)
X_train_poly = poly.fit_transform(X_train)
X_test_poly = poly.transform(X_test)
```

## Piplelining Concepts

```python
# WHatever degree you enter you will get the best fit line for that degree. You can also use cross validation to find the best degree for your model.

def poly_regression(degree):
    poly = PolynomialFeatures(degree=degree, include_bias=True)
    X_train_poly = poly.fit_transform(X_train)
    X_test_poly = poly.transform(X_test)
    
    regression = LinearRegression()
    regression.fit(X_train_poly, Y_train)
    y_pred = regression.predict(X_test_poly)
    score = r2_score(Y_test, y_pred)
    
    plt.scatter(X_train, regression.predict(X_train_poly), color='r')
    plt.scatter(X_train, Y_train)
    
    X_new = np.linspace(-3, 3, 200).reshape(200, 1)
    X_new_poly = poly.transform(X_new)
    y_new = regression.predict(X_new_poly)
    
    plt.plot(X_new, y_new, color='r', linewidth=3, label='New Predictions')
    plt.plot(X_train, Y_train, 'b.', label='Training Data')
    plt.plot(X_test, Y_test, 'g.', label='Testing Data')
    plt.xlabel('X Dataset')
    plt.ylabel('Y Dataset')
    plt.legend()
    plt.title(f'Polynomial Regression Degree: {degree}, R2 Score: {score:.4f}')
    plt.show()
```

```python
poly_regression(8)
```
