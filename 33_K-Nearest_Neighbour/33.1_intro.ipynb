## K Nearest Neighbour
- It is a supervised learning algorithm.
- used to solve both classification and regression problems.


## Clasification Problem

- try to find out from those k values , how many neighbours belongs to 0' s and 1's. (max neighbour will be output)




- we'll be using any distance based on model accuracy // for classification porblem

## Regression problem 



- Take the average of all the distance in regression

## nOW WE HAVE ISSUE REGARDING TIME COMPLEXITY

- for the curretn apprach the TC : O(n)
- for millions of data is will be an issue

> We use 1- KD Tree or 2 - Binary tree

## KD TREE



- Backtracking happens 

- It is a divide and conquer approach

- The TC is : O(log n)

## BALL TREE



- we form a group of nearest elements
- and then we find the nearest element in that group
- if k=4 , then we find 4 nearest elements inside the group which reduce the time significantly
