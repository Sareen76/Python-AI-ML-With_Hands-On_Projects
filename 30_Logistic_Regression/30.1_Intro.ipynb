## Logistic Regression

- used to predict the probability of an event happening
- used for binary classification problems
- uses sigmoid function to map the output to a probability between 0 and 1




## In the image :
    - the white line is the best fit line when there is no outlier in the data
    - the yellow line is the best fit line when there is an outlier in the data
    - we can see the line shift towards right so , for the data where the classiccation should be true becomes false and vice versa
    - also, there is a issue regarding the prediction value, it should lie bwtween 0 to1 , but in linear regression it could become greater than 1 and less than 0.

Why we can't use Linear regression for Classification ?
- outlier (best fit line changes)
- prediction value is not between 0 to 1. need squashing of values to 0 or 1.

> That's why we use logistic regression
