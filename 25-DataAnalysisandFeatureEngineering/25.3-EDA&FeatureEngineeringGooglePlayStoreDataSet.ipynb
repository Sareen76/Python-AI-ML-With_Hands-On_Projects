## Inisights and Obeservations

-->  The Dataset has missing values

```python
df.head()
```

```python
df[~df['Reviews'].str.isnumeric()]
#  We could see that 1 number is not numeric and will bring issue in model prediction .
```

```python
# Let's delete the record or convert it to million (let's delete it for now)

# Take a copy of dataset and do changes there
df_copy = df.copy()
```

```python
df_copy = df_copy.drop(df_copy.index[10472])
```

```python
df_copy[~df_copy['Reviews'].str.isnumeric()]
# Successfully dropped the data
```

```python
# Convert Review Datatype to int
df_copy['Reviews'] = df_copy['Reviews'].astype(int)
```

```python
df_copy.info()
```

```python
df_copy['Size'] = df_copy['Size'].str.replace('M', '000')
df_copy['Size'] = df_copy['Size'].str.replace('K', '')
df_copy['Size'] = df_copy['Size'].str.replace('k', '')
df_copy
```

```python
# We don't have any null values in Size Attribute and now all the values are numerical and we're not done yet

df_copy['Size'].isnull().sum()
```

```python
df_copy['Size'].unique()
```

```python
#  Now i can see there is an non-numerical values i.e , varies with device
# How did i got this??

df_copy['Size'] = df_copy['Size'].replace('Varies with device', np.nan)
df_copy['Size'] = df_copy['Size'].astype(float)
df_copy['Size'] 

# In Feature Engineering we will be changing the NAN value to something else
```

## Now handle the Installs Feature


```python
df_copy['Installs'].unique()
```

```python
df_copy['Price'].unique()
```

```python
chars_to_remove = ['+', ',', '$']
cols_to_clean = ['Installs', 'Price']
for item in chars_to_remove:
    for cols in cols_to_clean:
        df_copy[cols] = df_copy[cols].str.replace(item, '')
```

```python
df_copy['Price'].unique()
```

```python
# Now convert Price and Installs to Integer

df_copy['Installs'] = df_copy['Installs'].astype(int)
df_copy['Price'] = df_copy['Price'].astype(float)
df_copy.info()
```

```python
# Handling Last Updated feature

df_copy['Last Updated'] = pd.to_datetime(df_copy['Last Updated'])
df_copy['Day'] = df_copy['Last Updated'].dt.day
df_copy['Month'] = df_copy['Last Updated'].dt.month
df_copy['Year'] = df_copy['Last Updated'].dt.year

# df_copy.drop('Last Updated', axis=1,  inplace=True)
```

```python
df_copy.head()
```

```python
df_copy.to_csv('./CleanedData/google_cleaned.csv')
```
