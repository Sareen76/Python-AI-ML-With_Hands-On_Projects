```python
# Feature Engineering Properties
df['Date'] = df['Date_of_Journey'].str.split('/').str[0]  # Get the Dates of each data
df['Month'] = df['Date_of_Journey'].str.split('/').str[1]
df['Year'] = df['Date_of_Journey'].str.split('/').str[2]
```

```python
df['Date'] = df['Date'].astype(int)
df['Month'] = df['Month'].astype(int)
df['Year'] = df['Year'].astype(int)
df.info()
```

```python
# Drop the Date_of_Journey column as we don't require this object type value for our model

df.drop('Date_of_Journey',axis=1, inplace=True)
```

```python
df.head()
```

```python
df['Arrival_Time'] = df['Arrival_Time'].apply(lambda x:x.split(' ')[0])
```

```python
# Now we need to handle Arrival_Time column

df['Arrival_hour'] = df['Arrival_Time'].str.split(':').str[0].astype(int)
df['Arrival_min'] = df['Arrival_Time'].str.split(':').str[1].astype(int)

df.info()
df
```

```python
# Now drop the column Arrival_Time 

df.drop('Arrival_Time',axis = 1, inplace=True)
df.info()
```

```python
df.info()
```

```python
df['Dep_Hour'] = df['Dep_Time'].str.split(':').str[0].astype(int);
df['Dep_Min'] = df['Dep_Time'].str.split(':').str[1].astype(int);

df.drop('Dep_Time',axis = 1, inplace = True)
```

```python
df.info()
```

```python
df.head()
```

# Now lets Start with Categorical Features

```python
df['Total_Stops'].unique()
```

```python
df[df['Total_Stops'].isnull()]
```

```python
# Now replace the NAN Value to 1 stop

df['Total_Stops'] = df['Total_Stops'].map({'non-stop':0, '1 stop':1, '2 stops':2, '3 stops':3, '4 stops':4, np.nan:1})
```

```python
# We don't need the route , so remove the column

df.drop('Route', axis = 1, inplace=True)
```

```python
# Feature Engineering our Duration Column

df['Duration_Hour'] = df['Duration'].str.split(' ').str[0].str.split('h').str[0]
df['Duration_Min'] = df['Duration'].str.split(' ').str[1].str.split('m').str[0]

df.drop('Duration', axis= 1, inplace=True)

df
```

```python
df[df['Duration_Min'].isnull()]
```

```python
# Changing the null Value to 0

df['Duration_Min'] = df['Duration_Hour'].fillna(0)
```
