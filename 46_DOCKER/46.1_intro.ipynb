## WHY CONTAINER ??



- if developer A is working on windows and then developer B comes , then he have to install alll the dependencies, libraries etc . it will take time and any mismatch will result in error 

- let's say, both move to DEV Environment , and the project is deployed there for QA team in QA Server , then we have to install all the dependencies there also , it will take time and any mismatch will result in error

- if we have a container , then we can share the same environment for both developer and QA , and any mismatch will not result in error


Containers : It's a way to package applications with all the necessary dependecies and configuration

- pacakage is a portable artifact, easily share and move this package to any environment

- this makes the development and deployment more easy and efficient.

- The Docker image is preconfigured with the required Python version and toolset. QA teams only need to pull the image and launch the container to execute the application and perform validation.



## WHAT IS DOCKER THEN ?

## What Is Docker?

**Docker** is an open-source platform for building, packaging, shipping, and running applications in containers.

A **container** is a lightweight and portable unit that includes:

- The application
- Runtime environment
- Libraries and dependencies
- Configuration files

Containers ensure that applications run consistently across different environments.

### Docker vs. Virtual Machines

Unlike virtual machines, containers share the host operating system's kernel. This makes them:

- Faster to start
- More resource-efficient
- Easier to distribute and manage

> **Docker principle:** *Build once, run anywhere.*
