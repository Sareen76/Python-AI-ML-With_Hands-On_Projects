>FEW EXAMPLES AND APPLICATION OF LOGGING


```python
import logging

logging.basicConfig(
    level = logging.DEBUG,
    format = '%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datefmt = '%Y-%m-%d %H:%M:%S',
    handlers = [
        logging.FileHandler('application.log'),
        logging.StreamHandler()
    ]
)

# Call the module
logger = logging.getLogger('ArithmaticApp')

def add(a, b):
    result = a + b
    logger.debug(f'Adding {a} and {b} to get {result}')
    return result

def divide(a, b):
    try:
        result = a / b
        logger.debug(f'Dividing {a} by {b} to get {result}')
        return result
    except ZeroDivisionError:
        logger.error('Attempted to divide by zero')
        return None

def multiply(a, b):
    result = a * b
    logger.debug(f'Multiplying {a} and {b} to get {result}')
    return result

def subtract(a, b):
    try:
        result = a-b
        if result < 0:
            raise ValueError("Result is negative")
        logging.debug(f'Subtracting {b} from {a} to get {result}')
        return result
    except ValueError as ve:
        logger.error(f'Error in subtraction: {ve}')
        return None
add(5, 3)
divide(10, 1)
multiply(4, 2)
subtract(5, 10)

```
